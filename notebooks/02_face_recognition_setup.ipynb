# 02 - Face Recognition Setup (Haar Cascade + Encoding DB)

Builds the initial `face_db.pkl` from a folder of consenting-sample enrollment images (e.g. an LFW subset used for practice only), using the same Haar Cascade detection + histogram-encoding pipeline implemented in `app/services/cv_service.CVService`.

**Consent note:** only use images of individuals who have explicitly consented to biometric enrollment. See the Ethics & Privacy section of the project report.

```python
import sys
sys.path.append('..')
from app.services.cv_service import CVService
import glob, os

cv = CVService()
ENROLL_DIR = '../data/face_enrollment_samples'  # <customer_id>.jpg per file
```

```python
for path in glob.glob(os.path.join(ENROLL_DIR, '*.jp*g')):
    customer_id = os.path.splitext(os.path.basename(path))[0]
    with open(path, 'rb') as f:
        img_bytes = f.read()
    try:
        cid = cv.enroll_face(img_bytes, customer_id=customer_id)
        print(f'Enrolled {cid}')
    except ValueError as e:
        print(f'Skipped {path}: {e}')
```

```python
print(f'Total enrolled customers: {len(cv._face_db.encodings)}')
print('face_db.pkl persisted to app/models/face_db.pkl')
```
